**Note: - Since SVC is classifying the data very well. I would like to perform HyperParameter Tuning on it and get the best set of Hyperparameters.**

```python
import time
import pandas as pd

from sklearn.svm import SVC
from sklearn.model_selection import RandomizedSearchCV, StratifiedKFold

# Cross Validation
cv = StratifiedKFold(
    n_splits=10,
    shuffle=True,
    random_state=42
)

# Model
model = SVC(random_state=42)

# Hyperparameter Space
params = {
    'C': [0.1, 1, 10, 100, 1000],
    'gamma': [1, 0.1, 0.01, 0.001, 0.0001],
    'kernel': ['rbf']
}

# Random Search
search = RandomizedSearchCV(
    estimator=model,
    param_distributions=params,
    n_iter=10,
    cv=cv,

    # Multiclass metrics
    scoring={
        'accuracy': 'accuracy',
        'f1_macro': 'f1_macro',
        'f1_weighted': 'f1_weighted'
    },

    refit='f1_macro',

    n_jobs=-1,
    verbose=1,
    random_state=42,

    return_train_score=True
)

# Train
start_time = time.time()
search.fit(x_train, y_train)
end_time = time.time()
total_time = end_time - start_time

# Results
results = pd.DataFrame(search.cv_results_)

# Select useful columns
results = results[
    [
        'params',

        'mean_train_accuracy',
        'mean_test_accuracy',

        'mean_train_f1_macro',
        'mean_test_f1_macro',

        'mean_fit_time'
    ]
]

# Rename columns
results.columns = [
    'params',

    'train_accuracy',
    'cv_accuracy',

    'train_f1_macro',
    'cv_f1_macro',

    'training_time_sec'
]

# Overfitting indicators
results['accuracy_gap'] = (
    results['train_accuracy'] -
    results['cv_accuracy']
)

results['f1_gap'] = (
    results['train_f1_macro'] -
    results['cv_f1_macro']
)

# Sort by best validation F1
results = results.sort_values(
    by='cv_f1_macro',
    ascending=False
)

# Output
print("Best Parameters:")
print(search.best_params_)

print("\nBest Macro F1:")
print(search.best_score_)

print("\nTotal Search Time:")
print(round(total_time, 2), "seconds")

print("\nTop Results:")
print(results.head())

# Save
results.to_csv(
    "multiclass_svm_results.csv",
    index=False
)
```
